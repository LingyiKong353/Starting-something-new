# Chapter 2, Example 1: My first agent

The smallest possible agent: one set of instructions, no tools, plain-text output. It takes a research topic and returns a short research plan.

**Based on:** `chapter_02/01_first_agent.py` from [AI-Agent-Workflows](https://github.com/cxbxmxcx/AI-Agent-Workflows), the sample code for *AI Agents in Action, Second Edition* (Manning), Apache License 2.0. See [ATTRIBUTION.md](../../ATTRIBUTION.md).

**What I changed:** converted the script to a notebook. A notebook already runs an event loop, so I call `await Runner.run(...)` instead of `Runner.run_sync(...)`. I also renamed the variable `input` to `user_input` (it hid Python's built-in `input`), and `.env` is now searched for upward from the notebook's folder.

**To run it yourself:** put your `OPENAI_API_KEY` in a `.env` file in the repository root, install `requirements.txt` plus `ipykernel`, and run the cells in order.

## 1. Imports and API key

`load_dotenv` reads the `.env` file so the SDK can find `OPENAI_API_KEY`.

```python
# Adapted from chapter_02/01_first_agent.py in
# https://github.com/cxbxmxcx/AI-Agent-Workflows (Apache License 2.0).
# Modified: converted to a notebook; uses `await Runner.run(...)` instead of
# `Runner.run_sync(...)`; renamed `input` to `user_input`; .env is searched
# for upward from the notebook's folder.

from agents import Agent, Runner
from dotenv import find_dotenv, load_dotenv

# Load OPENAI_API_KEY (and other settings) from the .env file in the repo root
load_dotenv(find_dotenv(usecwd=True))
```

## 2. Define the agent

The `instructions` string is the agent's system prompt. It sets the role (research planning assistant), the task, and one hard constraint: exactly 5 tasks of 5 words or less. The agent has no tools and no special output type, so it just replies with text.

```python
# Agent Instructions
instructions = """
You are a research planning assistant.

**TASK INSTRUCTIONS**
- You will be given a research topic.
- Your task is to provide a plan on how to research this topic.
- Output 5 concise tasks (5 words or less) to your plan.
"""

agent = Agent(
    name="Research Planner",
    instructions=instructions,
)
```

## 3. Run it

`Runner.run` sends the input to the model and waits for the reply. `result.final_output` holds the agent's final answer.

*The output below was captured from my own run on 2026-10-27.*

```python
user_input = "learn about AI agents"

result = await Runner.run(agent, input=user_input)
print(result.final_output)
```

## What I noticed

- The agent followed the constraint: five tasks, each five words or fewer.
- The wording will differ from run to run, because the model's output is not deterministic.
- The instructions did all the work here. Later examples add model settings, structured output, tracing, and tools.
